```python
import numpy as np

def Raerod(Ta, ua, z1, z2, z0m, z0h, Ts, u0):
    """
    Compute turbulent aerodynamic resistance
    Analytical procedure in Louis (1979) and Mascart (1995)

    Parameters:
    Ta - Absolute virtual potential temperature in K (air temperature)
    ua - Wind velocity
    z1, z2 - Heights for the measurements
    z0m, z0h - Aerodynamic roughness lengths for momentum and heat
    Ts - Surface temperature
    u0 - Wind velocity at reference height

    Returns:
    Res - Aerodynamic resistance
    """
#z1 = Za, z2 = Zr(for RR) or d(for Rcan), z0m= ZmR(for RR) or Zmc(for Rcan), z0h= ZhR(for RR) or Zhc(for Rcan), Ts= TR(for RR) or Tcan(Rcan), u0= 0(for RR) or Us(for Rcan)
    # Constants
    g = 9.81           # gravity constant
    k = 0.35           # Von-Karman constant
    b = 9.4
    B = 9.4 / 2
    R = 0.74
    m = np.log(z0m / z0h)
    U = ua - u0
    z = z1 - z2

    # Using effective temperature to compute stability coefficient
    dT = Ta - Ts
    Tm = (Ta + Ts) / 2

    # Structure functions in Mascart et al (BLM 1995)
    a2 = k ** 2 / (np.log(z / z0m) ** 2)  # drag coefficient under neutral condition
    Ri = g * z * dT / (Tm * U ** 2)
    C = np.log(z / z0m) / np.log(z / z0h)
    Chs = 3.2165 + 4.3431 * m + 0.536 * m ** 2 - 0.0781 * m ** 3
    ph = 0.5802 - 0.1571 * m + 0.0327 * m ** 2 - 0.0026 * m ** 3
    Ch = Chs * a2 * b * C * (z / z0h) ** ph

    # Compute resistance and heat flux
    if Ri <= 0:
        Fh = C * (1 - b * Ri / (1 + Ch * np.sqrt(abs(Ri))))
    else:
        Fh = C / (1 + B * Ri) ** 2

    Res = R / (U * a2 * Fh)

    return Res
```
