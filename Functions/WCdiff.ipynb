```python
import numpy as np
import pandas as pd

# Define the function
def WCdiff(inflow, outflow, WC0, D, K, d, nL, dt):
    """
    Resolve 1D vertical water content diffusion process based on Richards' equation.

    Parameters:
    inflow = SWG, SWR : float - inflow boundary condition
    outflow = KRe(end),0 : float - outflow boundary condition
    WC0     : array - initial profile of water content
    D = DGe,DRe      : array - diffusion coefficient for each layer
    K = KGe,KRe     : array - hydraulic conductivity for each layer
    dgG     : array - thickness of vegetated ground layers
    nL      : int   - number of discretized layers
    dt      : float - time step used to resolve transport

    Returns:
    WCt = WRv,WGv : array - final profile of water content
    """

    # Initialize diffusion water content within layers
    DWG = np.zeros(nL)
    WCt = np.zeros(nL)

    # Update diffusive water transport within layers
    DWG[0] = 2 * D[0] * (WC0[0] - WC0[1]) / (d[0] + d[1]) + K[0]
    DWG[-1] = outflow

    for j in range(1, nL - 1):
        DWG[j] = 2 * D[j] * (WC0[j] - WC0[j + 1]) / (d[j] + d[j + 1]) + K[j]

    # Calculate the final water content profile
    WCt[0] = WC0[0] + dt * (inflow - DWG[0]) / d[0]

    for j2 in range(1, nL):
        WCt[j2] = WC0[j2] + dt * (DWG[j2 - 1] - DWG[j2]) / d[j2]

    return WCt


```
