```python
import numpy as np
from scipy.io import loadmat

# Constants
Rd = 287               # gas constant for dry air [J/kg/K]
Rv = 461.5             # gas constant for vapor
Lv = 2.26e6            # latent heat of vaporization [J/kg]

# qsat function to compute saturated specific humidity
def qsat(Lv, Rv, Rd, T, P):
    """
    Compute saturated specific humidity for given T and P
    using Clausius-Clapeyron Equation.

    Parameters:
    Lv : float
        Latent heat of vaporization [J/kg]
    Rv : float
        Gas constant for vapor [J/kg/K]
    Rd : float
        Gas constant for dry air [J/kg/K]
    T : numpy array
        Temperature [K]
    P : numpy array
        Atmospheric pressure [Pa]

    Returns:
    qs : numpy array
        Saturated specific humidity.
    """
    Tref = 298  # reference temperature @ 25°C
    eref = 3167  # reference saturated vapor pressure at Tref

    # Clausius-Clapeyron equation
    es = eref * np.exp(Lv * (T - Tref) / (Rv * T * Tref))
    rs = (Rd / Rv) * es / (P - es)
    qs = rs / (rs + 1)

    return qs

# Load atmospheric pressure (Pa_in) from the loaded .mat file
file_path = '/content/sample_data/Irri_input.mat'
mat_data = loadmat(file_path)

# Display the contents of the .mat file
mat_data.keys()
Pa_in = mat_data['Pa_in'].flatten()  # Flattening in case it's a column vector

# Define nd, np, and calculate Pa based on i >= 38
Nd = 6
np_value = 288 * Nd
i = 38  # Assuming i >= 38
Pa = Pa_in[(i-1) * np_value : i * np_value]  # Atmospheric pressure [Pa]

# Wall type setup
nW = 2  # Number of wall types: brick and glass
nt = len(Pa)  # Length of the Pa array (number of time steps)

# Initialize TW (wall temperatures) and generate random values for temperature
TW = np.random.uniform(273, 323, size=(nt, nW))  # Use your own temperatures in Kelvin for wall types

# Expand Pa to match the shape of TW for both wall types
Pa_expanded = np.tile(Pa, (nW, 1)).T  # Expand Pa to shape (nt, nW)

# Compute saturated specific humidity using the qsat function with TW
qs = qsat(Lv, Rv, Rd, TW, Pa_expanded)

# Display the result
qs
```
