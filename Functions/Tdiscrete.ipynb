```python
import numpy as np

def Tdiscrete(dt, qW, GW, dW, cW, nL, nW, TW0):
    """
    Compute solid temperatures for walls, roofs, or roads over time.

    Parameters:
    dt : float
        Time step [s].
    qW : numpy array
        Net heat flux at the wall/roof surface [W/m²].
    GW : numpy array
        Conductive heat flux between layers [W/m²].
    dW : numpy array
        Thickness of each layer [m].
    cW : numpy array
        Heat capacity of each layer [J/kg/K].
    nL : int
        Number of layers in the wall or roof.
    nW : int
        Number of wall/roof types.
    TW0 : numpy array
        Initial temperatures of the layers [°C].

    Returns:
    TW : numpy array
        Updated temperature of the layers after one time step [°C].
    """
    TW = np.zeros((1, nL, nW))

    # Compute temperature for the first layer
    for i in range(nW):
        TW[0, 0, i] = TW0[0, 0, i] + dt * (qW[0, i] - GW[0, 0, i]) / dW[0, i] / cW[0, i]

    # Compute temperature for subsequent layers
    for j in range(1, nL):
        for i in range(nW):
            TW[0, j, i] = TW0[0, j, i] + dt * (GW[0, j - 1, i] - GW[0, j, i]) / dW[j, i] / cW[j, i]

    return TW

# Example usage:
dt = 3600  # Time step (1 hour in seconds)
nL = 3  # Number of layers
nW = 2  # Number of wall types

# Example input data
qW = np.array([[500, 450]])  # Net heat flux for 2 wall types (W/m²)
GW = np.array([[[150, 100], [120, 90], [80, 70]]])  # Conductive heat flux between layers (W/m²)
dW = np.array([[0.2, 0.15], [0.1, 0.1], [0.05, 0.05]])  # Thickness of layers (m)
cW = np.array([[800, 850], [900, 900], [1000, 950]])  # Heat capacity of materials (J/kg/K)
TW0 = np.array([[[20, 18], [19, 17], [18, 16]]])  # Initial temperatures (°C)

# Call the function
TW = Tdiscrete(dt, qW, GW, dW, cW, nL, nW, TW0)

TW # Display updated temperatures
```
